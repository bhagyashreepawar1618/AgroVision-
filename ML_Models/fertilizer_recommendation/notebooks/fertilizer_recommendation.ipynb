## Exploring Data (Visualization)

```python
plt.figure(figsize = (8, 5))
sns.countplot(x = 'Fertilizer Name', data = df, order = df['Fertilizer Name'].value_counts().index)
plt.xticks(rotation = 45)
plt.title('Fertilizer Distribution')
plt.ylabel('Count')
plt.show()
```

### Insights
- Urea is the most common fertilizer (22 of 99 rows), 10-26-26 and 17-17-17 are the rarest (7 each)
- The dataset is moderately imbalanced — not extreme, but the model will see roughly 3x more Urea examples than the rarest classes, so it'll likely be more confident predicting Urea than the rare ones

```python
plt.figure(figsize = (10, 6))
sns.countplot(x = 'Soil Type', hue = 'Fertilizer Name', data = df)
plt.title('Fertilizer Distribution by Soil Type')
plt.legend(bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.show()
```

### Insights
- Black soil → Urea dominates (7 of ~19 rows for that soil)
- Red soil → DAP dominates (6 of ~19 rows)
- Sandy, Loamy, Clayey → no clear single winner, fairly mixed
- Takeaway: soil type gives the model a strong hint for Black/Red, a weak hint otherwise

```python
plt.figure(figsize = (14, 8))
sns.countplot(x = 'Crop Type', hue = 'Fertilizer Name', data = df)
plt.xticks(rotation = 45)
plt.title('Fertilizer Distribution by Crop Type')
plt.legend(bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.show()
```

### Insights
- Millets → Urea leads; Sugarcane → 20-20 leads; Cotton → DAP leads; Ground Nuts → DAP leads
- But each crop only has 6-13 total rows, so these "leading" fertilizers are often based on just 3-4 data points — much shakier evidence than the soil-type patterns
- Takeaway: crop type is a weaker, noisier signal than soil type in this dataset

```python
plt.figure(figsize = (10, 6))
sns.countplot(x = 'Soil Type', hue = 'Crop Type', data = df)
plt.title('Crop Distribution by Soil Type')
plt.legend(bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data = df, x = 'Fertilizer Name', y = 'Nitrogen')
plt.xticks(rotation = 45)
plt.title('Nitrogen Level by Fertilizer')
plt.show()
```

#### Does temperature vary across fertilizer recommendations?

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data = df, x = 'Fertilizer Name', y = 'Temperature')
plt.xticks(rotation = 45)
plt.title('Temperature Level by Fertilizer')
plt.show()
```

- Most fertilizers have a median temperature between 29°C and 31°C.
- This suggests fertilizers are generally recommended within a similar temperature range.
